# Pruning and Truncation in Pauli Propagation

This tutorial introduces the pruning and truncation strategies in `pprop` and shows how they reduce the cost of Heisenberg-picture Pauli propagation.

We use the 2D transverse-field Ising model on a $8 \times 8$ lattice as the running example.

## What is pruning?

During Heisenberg evolution the `PauliDict` can grow large as each gate splits Pauli words into multiple branches. Many of these branches are guaranteed to contribute zero to the final expectation value

$$\langle 0 | O | 0 \rangle$$

because they can never reach the $Z/I$ subspace required by `to_expectation`. Pruning detects and discards these branches **exactly**, the result is identical to unoptimised propagation, only faster.

`pprop` provides two exact pruning strategies, both in `pprop.propagator.pruning`:

- **`DeadQubitPruner`**: removes any Pauli word that carries $X$ or $Y$ on a qubit that will never be touched by any remaining gate. Such a qubit is permanently frozen in a non-$ZI$ state.
- **`XYWeightPruner`**: removes any Pauli word whose XY-weight (number of qubits carrying $X$ or $Y$) exceeds the maximum reduction achievable by all remaining gates in the causal cone of that word.

## What is truncation?

Truncation strategies drop terms that exceed a budget or fall below a magnitude threshold. Unlike pruning, this is **approximate**, the result changes, but the `PauliDict` stays small and propagation stays fast.

`pprop` provides three truncation strategies in `pprop.propagator.truncation`:

- **`WeightTruncation(k1)`**: removes Pauli words whose weight (number of non-identity operators) exceeds `k1` after each gate step.
- **`FrequencyTruncation(k2)`**: removes coefficient terms whose total trigonometric frequency (total number of sin/cos factors) exceeds `k2` after each gate step.
- **`CoefficientTruncation(threshold)`**: removes individual coefficient terms whose scalar magnitude falls below `threshold`, more effective on non-parametrized circuits.

The legacy `k1` and `k2` parameters on `Propagator` are a shorthand for `WeightTruncation(k1)` and `FrequencyTruncation(k2)` respectively.

## Step 0: Imports

```python
import time

import numpy as np
import pennylane as qml

from pprop import Propagator
from pprop.propagator.pruning import DeadQubitPruner, XYWeightPruner
from pprop.propagator.truncation import (
    CoefficientTruncation, #noqa
    FrequencyTruncation,
    WeightTruncation,
)
```

## Step 1: Problem setup

We reuse the Hamiltonian and ansatz from the VQE tutorial. The Hamiltonian is

$$H(J, h) = -\frac{1}{N}\left(J\sum_{\langle i,j \rangle} Z_i Z_j + h\sum_i X_i\right)$$

on a $8 \times 8$ lattice with open boundary conditions, $J = h = 1$, and row-major qubit indexing.

```python
side: int = 4

J: float = 1.0
h: float = 1.0
num_qubits: int = side * side


def hamiltonian(side: int, J: float, h: float) -> qml.Hamiltonian:
    coeffs = []
    obs = []
    N = side * side
    for x in range(side):
        for y in range(side):
            i = x * side + y
            if y < side - 1:
                j = x * side + (y + 1)
                coeffs.append(-J / N)
                obs.append(qml.PauliZ(i) @ qml.PauliZ(j))
            if x < side - 1:
                j = (x + 1) * side + y
                coeffs.append(-J / N)
                obs.append(qml.PauliZ(i) @ qml.PauliZ(j))
    for i in range(N):
        coeffs.append(-h / N)
        obs.append(qml.PauliX(i))
    return qml.Hamiltonian(coeffs, obs)


def circuit(params):
    index = 0
    for q in range(num_qubits):
        qml.RX(params[index], wires=q); index += 1
    for d in range(2):
        y_start = 0 if d % 2 == 0 else 1
        for x in range(side):
            for y in range(y_start, side - 1, 2):
                qml.CNOT(wires=[x * side + y, x * side + (y + 1)])
    for q in range(num_qubits):
        qml.RX(params[index], wires=q); index += 1
    for d in range(2):
        x_start = 0 if d % 2 == 0 else 1
        for y in range(side):
            for x in range(x_start, side - 1, 2):
                qml.CNOT(wires=[x * side + y, (x + 1) * side + y])
    for q in range(num_qubits):
        qml.RY(params[index], wires=q); index += 1
    return qml.expval(hamiltonian(side, J, h))
```

## Step 2: Propagation with and without pruning

Passing the pruners in `pruners=[ ... ]` to `.propagate()` enables the specified pruning strategies. We time both runs to measure the speedup.

```python
prop_no_opt = Propagator(circuit)

t0 = time.perf_counter()
prop_no_opt.propagate(pruners=[])
t_no_opt = time.perf_counter() - t0

print(f"Time without pruning : {t_no_opt:.3f} s")

prop_opt = Propagator(circuit)

t0 = time.perf_counter()
prop_opt.propagate(pruners = [XYWeightPruner(), DeadQubitPruner()])
t_opt = time.perf_counter() - t0

print(f"Time with pruning    : {t_opt:.3f} s")
print(f"Speedup              : {t_no_opt / t_opt:.2f}x")
```

## Step 3: Correctness check

Pruning must never change the result. We evaluate both propagators at the same random parameter point and assert that the outputs agree.

```python
rng = np.random.default_rng(42)
params = rng.random(prop_opt.num_params)

e_no_opt = prop_no_opt(params)[0]
e_opt    = prop_opt(params)[0]

print(f"Without pruning : {e_no_opt}")
print(f"With pruning    : {e_opt}")

assert np.isclose(e_no_opt, e_opt, atol=1e-10), "Pruning changed the result"

print("\nOutputs match.")
```

```python
print(f"Optimized  : {len(prop_opt.paulidicts[0])}")
print(f"Unoptimized: {len(prop_no_opt.paulidicts[0])}")
```

## Step 4: Weight and frequency truncation

`WeightTruncation(k1)` and `FrequencyTruncation(k2)` are the explicit-object equivalents of the legacy `k1`/`k2` constructor arguments.

```python
prop_trunc = Propagator(circuit)

prop_trunc.propagate(truncations=[WeightTruncation(8), FrequencyTruncation(10)], pruners = [XYWeightPruner(), DeadQubitPruner()])
e_trunc  = prop_trunc(params)[0]


print(np.abs(e_no_opt - e_trunc))
```
